# 4.27 — Creating Pandas Series from Lists and Arrays

This notebook covers creating Pandas Series from Python lists and NumPy arrays.

## 1. Import Required Libraries

Import the necessary libraries, including Pandas and NumPy.

```python
import pandas as pd
import numpy as np
```

## 2. Understanding Pandas Series

A Pandas Series is a one-dimensional labeled array capable of holding any data type. It is like a column in a DataFrame or a labeled NumPy array.

Key points:
- It has data (values) and labels (index)
- Default index is 0, 1, 2, ...
- Values can be accessed by position or label

```python
# Example of a simple Series
data = [10, 20, 30, 40]
series = pd.Series(data)
print("Series:")
print(series)
print("\nIndex:", series.index)
print("Values:", series.values)
```

## 3. Creating a Series from Python Lists

You can create a Pandas Series directly from a Python list. The Series will automatically assign a default integer index starting from 0.

```python
# Creating a Series from a list
grades = ['A', 'B', 'C', 'D']
grades_series = pd.Series(grades)
print("Grades Series:")
print(grades_series)

# Numeric list
numbers = [1, 2, 3, 4, 5]
numbers_series = pd.Series(numbers)
print("\nNumbers Series:")
print(numbers_series)
```

## 4. Creating a Series from NumPy Arrays

You can also create a Series from a NumPy array. The data types are preserved, and the Series adds labels to the array.

```python
# Creating a Series from a NumPy array
array = np.array([1.1, 2.2, 3.3, 4.4])
array_series = pd.Series(array)
print("Array Series:")
print(array_series)
print("Data type:", array_series.dtype)

# Compare with raw NumPy array
print("\nRaw NumPy array:")
print(array)
print("Array dtype:", array.dtype)
```

## 5. Understanding Index and Values

The Series has two main components: the index (labels) and the values (data). You can access them separately.

```python
# Inspecting index and values
sample_series = pd.Series([100, 200, 300], index=['a', 'b', 'c'])
print("Sample Series with custom index:")
print(sample_series)

print("\nIndex:")
print(sample_series.index)

print("\nValues:")
print(sample_series.values)

# Accessing by label vs position
print("\nAccess by label 'b':", sample_series['b'])
print("Access by position 1:", sample_series.iloc[1])
```

## 6. Performing Simple Operations on a Series

You can perform various operations on a Series, such as arithmetic, slicing, and more.

```python
# Simple operations
nums = pd.Series([1, 2, 3, 4, 5])
print("Original Series:")
print(nums)

print("\nSlicing (first 3):")
print(nums[:3])

print("\nArithmetic (add 10):")
print(nums + 10)

print("\nMean:", nums.mean())
print("Sum:", nums.sum())
```
