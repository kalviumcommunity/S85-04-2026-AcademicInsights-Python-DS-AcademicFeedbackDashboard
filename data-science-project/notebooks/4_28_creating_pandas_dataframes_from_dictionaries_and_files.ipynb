# 4.28 — Creating Pandas DataFrames from Dictionaries and Files

This notebook covers creating Pandas DataFrames from Python dictionaries and files.

## 1. Import Required Libraries

Import the necessary libraries, including Pandas.

```python
import pandas as pd
import numpy as np
```

## 2. Understanding Pandas DataFrames

A DataFrame is a 2-dimensional labeled data structure with columns of potentially different types. It's like a spreadsheet or SQL table.

```python
# Example DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['NYC', 'LA', 'Chicago']
}
df = pd.DataFrame(data)
print("DataFrame:")
print(df)
```

## 3. Creating DataFrames from Dictionaries

You can create a DataFrame from a dictionary where keys are column names and values are lists of data.

```python
# Creating DataFrame from dictionary
student_data = {
    'Student': ['John', 'Jane', 'Mike'],
    'Math': [85, 92, 78],
    'Science': [88, 95, 82]
}
df_students = pd.DataFrame(student_data)
print("Students DataFrame:")
print(df_students)
```

## 4. Creating DataFrames from Files

You can load data from CSV files into DataFrames using pd.read_csv().

```python
# Loading DataFrame from CSV
df_csv = pd.read_csv('../data/sample_data.csv')
print("DataFrame from CSV:")
print(df_csv)
```

## 5. Inspecting DataFrame Structure

Use methods like head(), info(), and shape to inspect the DataFrame.

```python
# Inspecting DataFrame
print("First 5 rows:")
print(df_csv.head())

print("\nDataFrame info:")
print(df_csv.info())

print("\nShape:", df_csv.shape)
print("Columns:", list(df_csv.columns))
```
